Download the data size 64, 128 and 224

```python
from pathlib import Path
from medmnist import INFO
from torchvision.datasets.utils import download_url

DATA = Path(
    r"C:\Users\XiXi\OneDrive\Documents\Stanford\CS230\pathmnist_project\data"
)
DATA.mkdir(parents=True, exist_ok=True)

info = INFO["pathmnist"]

# download all three sizes in PathMnist
for size in [64, 128, 224]:
    print(f"\nDownloading {size}×{size} images")
    download_url(
        url=info[f"url_{size}"],
        root=str(DATA),
        filename=f"pathmnist_{size}.npz",
        md5=info[f"MD5_{size}"],
    )
```

```python
for file in DATA.glob("*.npz"):
    print(f"{file.name}: {file.stat().st_size / 1024**3:.2f} GB")
```
